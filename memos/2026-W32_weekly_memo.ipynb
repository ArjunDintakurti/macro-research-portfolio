# A steepener that's holding its shape

*Weekly note — week ending 07 August 2026*

US yields eased back a little this week — the 10-year slipped from 4.75% to 4.65% — which on its own could read as last week's move partially reversing. What's actually more informative is that 2s10s barely changed despite that pullback, holding around 0.46 percentage points versus 0.47 the week before. The front end (the 2-year yield) eased by almost exactly as much as the 10-year did, so the curve kept its shape even as the general level of yields drifted lower.

That distinction matters because a curve that holds its steepness through a modest rally is a better signal of a durable regime change than a single sharp widening on its own would be — it suggests the market's underlying view (gradual normalisation, not recession) survived a week where the initial move partly unwound, rather than only existing because of one outsized data point.

The yen, for its part, kept strengthening — USD/JPY fell again, from just above 160 to just above 158, confirming last week's reversal wasn't a one-off overshoot. Four weeks of carry-trade-driven weakness gave way to two weeks of correction, and at this point the pair is essentially back to where it started this series in early June — a reminder that a month-long trend and a two-week reversal can end up cancelling each other out almost exactly, which is easy to forget in the middle of either move.

The next real test for the steepening view is whichever US data print comes next — the curve holding its shape through a quiet data week is reassuring, but it hasn't yet been tested against a print that actually surprises the market.
