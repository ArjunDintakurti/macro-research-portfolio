## This week's macro backdrop

Two things happened this week, in opposite directions, and both matter. First, US yields kept climbing — 10Y to 4.75%, 30Y to a series-high 5.27% — and this time the curve actually steepened with it: 2s10s widened from +0.36pp to +0.47pp, the sharpest weekly widening in this series so far. That's the steepener thesis from week one finally showing up clearly in the data rather than just holding its ground.

Second, USDJPY reversed hard — from 163.82 down to 160.24, a roughly 2.2% yen rally in a single week, the largest single-week FX move in this entire series. That's a sharp move against the short-JPY leg opened last week, and it happened fast enough that the position should already be closed by the invalidation rule set last week (161.00), not held in the hope of a bounce.

## Trade expression

- **Instrument / trade:** hold and add modestly to the 2s10s steepener, now working; close the short-JPY leg — the invalidation level was hit
- **Direction & size (illustrative):** steepener increased from small to small-plus, given the position is now working and the curve move looks structural rather than one-off; JPY leg closed at a loss
- **Entry context:** 2s10s at +0.47pp, widest print of the series
- **Time horizon:** steepener extended to 6-8 weeks given the trend is now established
- **Catalyst(s) to watch:** whether the curve continues to steepen or whether this week was itself the noise — one good week doesn't yet prove a trend either, which is worth remembering after just having been burned doing the opposite on JPY

## Risk section

- **What would prove this view wrong:** a snap-back flattening in 2s10s over the next 1-2 weeks would suggest this week's widening was itself the outlier, not the trend
- **Key risks to the trade:** the JPY loss this week is a useful, uncomfortable reminder that a rate-differential trade can be right on the fundamentals and still lose money on timing — that's worth sitting with rather than rationalising away
- **Invalidation level / stop condition:** 2s10s back below +0.35pp (last month's range) would be treated as the steepener thesis breaking down
