# The yen's quiet drift is starting to add up

*Weekly note — week ending 17 July 2026*

US rates had another unremarkable week — the curve barely moved, continuing the pattern from the last fortnight — so this note leans on the one part of the market that's actually been drifting in one direction for three straight weeks now: the yen.

USD/JPY has crept from just above 161 to just over 162 across this stretch, which on its own is not a large move by FX standards. What makes it worth a mention is the consistency: this isn't one sharp jump, it's three consecutive weeks in the same direction, and it's happening without any obvious single news event behind it. The more mundane explanation is usually the right one in cases like this — Japan has kept interest rates near zero for so long that the yen is the standard "funding" currency for carry trades: borrow cheap in yen, invest the proceeds somewhere that pays more, and the act of doing that at scale tends to weaken the yen as a byproduct. A widening gap between US and Japanese short rates, even a gradual one, is consistent with more of that trade being put on.

The catch with carry trades, and it's worth saying plainly rather than saving for a "risks" section at the bottom: they tend to unwind quickly and painfully rather than gradually, usually around a moment that makes the funding side of the trade look suddenly riskier — a Bank of Japan policy surprise, most obviously. Nothing this week points to that happening imminently, but a slow, one-directional grind like this is exactly the kind of move that's worth watching rather than dismissing as noise, precisely because of how these trades tend to end.
