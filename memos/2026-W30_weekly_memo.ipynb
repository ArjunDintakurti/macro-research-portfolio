# The first week that actually moved

*Weekly note — week ending 24 July 2026*

After three weeks of relative calm, this is the first genuinely active week in this series. The 10-year Treasury yield jumped by roughly 14 basis points over five trading days — a meaningful move by the standards of the past month — and the 30-year followed it up to a fresh high for this stretch, north of 5.15%. USD/JPY carried on drifting weaker too, touching its highest point of the summer so far, extending the pattern flagged last week.

The useful thing to notice here is what *didn't* change: 2s10s barely budged, because the 2-year moved up by roughly as much as the 10-year did. That's a parallel shift, not a steepening or a flattening — the market repriced the general level of US rates higher (perhaps trimming how many further cuts it expects this year) without changing its underlying view on the shape of the cycle. It's a genuinely different kind of move from the ones that tend to make headlines, and worth distinguishing from a steepening trade specifically, because the two imply quite different things about what the market believes is coming.

The yen move is the more interesting thread to keep pulling on. Three weeks of drift have now become four, and the level (above 163) is meaningfully weaker than where this series started in early June. Wider US-Japan rate differentials are the most obvious explanation, and this week's rates move — the US repricing higher — only widens that gap further. If there's a single thing worth watching most closely heading into next week, it's whether this yen weakness is a trend with further to run, or a stretched move due a correction.
