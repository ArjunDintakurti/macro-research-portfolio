This is the first week where the data actually moves enough to matter. ## Trade expression

- **Instrument / trade:** hold the 2s10s steepener (parallel shift this week, thesis unaffected); add a new, separate, small short-JPY-vs-USD position given the now-two-week trend of yen weakness alongside widening yield differentials
- **Direction & size (illustrative):** steepener unchanged; new JPY leg small and explicitly tactical
- **Entry context:** USDJPY at a series high (163.82), rate-differential story intact
- **Time horizon:** steepener 4-6 weeks (unchanged); JPY leg shorter, 2-3 weeks — carry trades can unwind fast and this is being sized and timed accordingly
- **Catalyst(s) to watch:** any Bank of Japan commentary (a hawkish surprise from the BoJ is the single biggest risk to a short-JPY position), plus continued US yield direction

## Risk section

- **What would prove this view wrong:** for the steepener, re-inversion; for JPY, any reversal back toward 160 would question whether the trend is real or just noise around a still-small sample
- **Key risks to the trade:** JPY carry trades are a classic "grind up slowly, unwind violently" pattern — this is exactly the kind of position that needs a firm invalidation level, not a "wait and see"
- **Invalidation level / stop condition:** USDJPY back below 161.00 on a closing basis would be treated as invalidating the JPY leg specifically
