## This week's macro backdrop

The pattern broke this week, and sharply. The 2Y yield jumped from 4.24% to 4.34% — a 15bp move, the largest front-end move since the series began — while the 10Y actually eased slightly (4.74%→4.73%). That combination compressed 2s10s hard, from +0.50pp down to +0.39pp, the steepener's biggest one-week reversal to date. A front end that rises while the back end doesn't move is the textbook signature of the market pricing in fewer near-term rate cuts, not of a broader risk-off move (which would typically hit the long end too, via a flight to quality).

GBPUSD and EURUSD both pulled back from last week's highs (1.1643, 1.3583) — consistent with the same story, since a hawkish US repricing tends to firm the dollar broadly, not just against one currency.
