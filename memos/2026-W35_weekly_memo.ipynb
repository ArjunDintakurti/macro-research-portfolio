# The steepener's first real reversal

*Weekly note — week ending 28 August 2026*

After four weeks of a steepening or stable US curve, this week broke the pattern, and broke it sharply. The 2-year yield jumped by roughly 15 basis points — the largest front-end move anywhere in this series — while the 10-year barely moved at all. That combination compressed 2s10s from around 0.50 percentage points down to 0.39, the steepener's biggest single-week reversal to date.

The shape of that move is the tell. A front end that rises sharply while the back end holds still is the textbook signature of the market pricing in fewer near-term Fed rate cuts, rather than a broader risk-off move — a genuine flight-to-quality episode would typically drag the long end down too, via demand for the safest, longest-dated government debt available. This looks more specific than that: a repricing of the near-term policy path, not a shift in growth or recession expectations.

Sterling and the euro both pulled back from last week's highs, which fits the same story — a hawkish US repricing tends to firm the dollar broadly rather than moving one pair in isolation. With the Fed's September decision now only a few weeks out, it's a reasonable moment to treat this week's move as the market starting to position for that meeting rather than a stray, unrelated data point — though one sharp week isn't yet enough to say the four-week steepening trend has properly reversed rather than just paused.
