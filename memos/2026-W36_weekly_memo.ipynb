The front-end repricing from last week continued rather than reversed: 2Y up again to 4.37% (from 4.34%), 10Y also higher at 4.78%, so 2s10s ticked up slightly to +0.41pp — still well below the August highs, but not the outright flattening some of last week's data might have suggested either. A simultaneous curve-flattening-at-the-long-end and a sharp yen rally, two weeks before a Fed decision, is a classic pre-FOMC positioning pattern: markets trimming risk and unwinding carry trades ahead of an event rather than expressing a fresh directional view.

- **Direction & size (illustrative):** small and explicitly tactical, sized for a two-week holding period bridging the FOMC

## Risk section

- **What would prove this view wrong:** a dovish FOMC surprise (more cuts signalled than priced) would likely re-steepen the curve and weaken the yen back, the opposite of this week's positioning
- **Key risks to the trade:** holding a position through a known, scheduled event is a deliberate choice to accept event risk rather than avoid it — that's a real risk, not a hidden one, and sizing is kept small because of it
- **Invalidation level / stop condition:** 5s30s back above +0.75pp would be treated as the flattening thesis failing
