# A curve that's stopped worrying

*Weekly note — week ending 03 July 2026*

Three years ago, an inverted 2s10s spread was the line everyone in this market watched. This is for good reason as it had preceded every US recession since the 1970s. This week it's worth noting what the curve is *not* doing: it's not inverted, it hasn't been for a while, and it's actually settled into a comfortable, modestly positive shape, with the 10-year yield sitting a little over a third of a percentage point above the 2-year.

That matters more than it might sound. The 2-year yield is mostly a bet on where the Fed funds rate will average over the next two years, while the 10-year adds a longer-run growth and inflation view on top. A positive, stable spread is the market's way of saying the current easing cycle — the effective funds rate has come down from a peak above 5% to a little under 3.7% — looks like an orderly return to normal rather than a scramble to head off a downturn. Inflation is still running above the Fed's 2% target, a touch above 3.5% on the latest reading, so this isn't a victory lap either. It's a curve that's stopped pricing panic in either direction, which is itself the story this week, however undramatic that sounds.

Nothing in the FX space this week broke from that theme. Sterling and the euro are both sitting close to where they started the month against the dollar, and the yen has drifted a little weaker as the rate gap between the US and Japan stays wide — more a slow grind than anything worth a headline yet.

Worth watching over the next few weeks: whether this shape holds as more data comes in, or whether it was just a quiet start to the summer before something moves it.
